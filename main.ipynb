```python
from bubble_sort import bubbleSort
from selection_sort import selectionSort
from insertion_sort import insertionSort
from random_array import generate_random_ints
import time
```

```python
array_lengths = [10, 1000, 5000, 10000,20000] # x axis values for the graph
# y values for the 3 graphs:
bubble_times = []
selection_times = []    
insertion_times = []
```

```python
now = time.time()
for array_length in array_lengths:
    random_array = generate_random_ints(array_length)

    # Bubble Sort
    bubble_array = random_array.copy()
    now = time.time()
    bubbleSort(bubble_array)
    bubble_times.append(time.time() - now)
    print(f"Bubble Sort took {bubble_times[-1]:.6f} seconds for array of length {array_length}")

    # Selection Sort
    selection_array = random_array.copy()
    now = time.time()

    selectionSort(selection_array)
    selection_times.append(time.time() - now)
    print(f"Selection Sort took {selection_times[-1]:.6f} seconds for array of length {array_length}")

    # Insertion Sort
    insertion_array = random_array.copy()
    now = time.time()

    insertionSort(insertion_array)
    insertion_times.append(time.time() - now)
    print(f"Insertion Sort took {insertion_times[-1]:.6f} seconds for array of length {array_length}")
```
