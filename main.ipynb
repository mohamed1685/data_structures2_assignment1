```python
from n_squared_sorting.bubble_sort import bubbleSort
from n_squared_sorting.selection_sort import selectionSort
from n_squared_sorting.insertion_sort import insertionSort
from nlogn_sorting.quick_sort import quickSort
from nlogn_sorting.merge_sort import mergeSort
from nlogn_sorting.heap_sort import heapSort
from hybrid_sorting.hybrid_sort import hybrid_sort
from random_array import generate_random_ints
import time
```

```python
array_lengths = [1000, 5000, 10000,20000,50000,100000] # x axis values for the graph
# y values for the graphs:
bubble_times = []
selection_times = []    
insertion_times = []
quick_times = []
merge_times = []
heap_times = []
hybrid_times = []
threshold = 6
```

```python
now = time.time()
for array_length in array_lengths:
    print(f"Array Length {array_length}")
    random_array = generate_random_ints(array_length)

    # Bubble Sort
    bubble_array = random_array.copy()
    now = time.time()
    bubbleSort(bubble_array)
    bubble_times.append(time.time() - now)
    print(f"Running time for bubbble sort is {bubble_times[-1]*1000}ms")

    # Selection Sort
    selection_array = random_array.copy()
    now = time.time()

    selectionSort(selection_array)
    selection_times.append(time.time() - now)
    print(f"Running time for selection sort is {selection_times[-1]*1000}ms")

    # Insertion Sort
    insertion_array = random_array.copy()
    now = time.time()

    insertionSort(insertion_array)
    insertion_times.append(time.time() - now)
    print(f"Running time for insertion sort is {insertion_times[-1]*1000}ms")

    # Quick Sort
    quick_array = random_array.copy()
    now = time.time()

    quickSort(quick_array, 0, len(quick_array) - 1)
    quick_times.append(time.time() - now)
    print(f"Running time for quick sort is {quick_times[-1]*1000}ms")

    # Merge Sort
    merge_array = random_array.copy()
    now = time.time()

    mergeSort(merge_array, 0, len(merge_array) - 1)
    merge_times.append(time.time() - now)
    print(f"Running time for merge sort is {merge_times[-1]*1000}ms")

    # Heap Sort
    heap_array = random_array.copy()
    now = time.time()

    heapSort(heap_array)
    heap_times.append(time.time() - now)
    print(f"Running time for heap sort is {heap_times[-1]*1000}ms")

    # Hybrid Sort
    hybrid_array = random_array.copy()
    now = time.time()

    hybrid_sort(hybrid_array, 0, len(hybrid_array) - 1, threshold)
    hybrid_times.append(time.time() - now)
    print(f"Running time for hybrid sort is {hybrid_times[-1]*1000}ms")
```
